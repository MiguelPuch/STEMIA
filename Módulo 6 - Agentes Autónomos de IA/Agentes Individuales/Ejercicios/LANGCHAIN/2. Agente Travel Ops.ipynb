# Agente “Travel Ops” con búsqueda web, tool-calling y confirmación

## Objetivo

Construir un agente de viajes que:

* tenga sesgo “amable” hacia Bilbao (como en tu material),

* use una herramienta de búsqueda (p. ej., Tavily) vía function calling de OpenAI,

* aplique un hook de confirmación antes de acciones “caro/masivas” (p. ej., consultar 3+ rutas),

* tracee la interacción con OTLP hacia LangSmith (opcional).

### Objetivo 1

* Implementa travel_search(query: str) -> str que devuelva contexto sintético (texto) a partir de la API elegida.

* Maneja cuotas, timeouts y errores de red.

* Devuelve un contexto condensado (limpia HTML, elige 3–5 snippets).

### Objetivo 2

* Define un SYSTEM_PROMPT con el sesgo “Bilbao fan”.

* Configura client.chat.completions.create(..., tools=[...], tool_choice="auto").

* Si el modelo solicita la tool, ejecuta la función y reenvía el resultado como mensaje role="tool".

### Objetivo 3

* Implementa un confirmation hook (p. ej., con rich.Prompt) que pregunte antes de:

    - realizar > N búsquedas en paralelo,

    - o cuando el coste/latencia estimada supere un umbral configurable.

* Si el usuario rechaza, aborta con un mensaje claro.

### Objetivo 4

* Configura TracerProvider + SimpleSpanProcessor(OTLPSpanExporter(...)).

* Emite spans para:

    - llamada al LLM,

    - ejecución de tool(es),

    - consolidación de respuesta.

* Propaga atributos (latencia, nº resultados, longitud del prompt).

### Objetivo 5


* Incluye escenarios de prueba:

    - “¿Qué trenes hay de Santander a Madrid?”,

    - “Busca vuelos para este fin de semana” (debe disparar el hook si N alto),

    - “Propón algo en el norte (guiño Bilbao)”.

* README con ejemplos y limitaciones.
