# Reportero tech con Hacker News + memoria + fallback offline

## Objetivo

Crear un reportero tecnológico que:

* obtenga titulares de Hacker News con una tool HTTP,

* exponga streaming,

* use memoria para recordar la sesión del editor,

* tenga caché/fallback local para cuando la red falle,

* (opcional) emita trazas OTLP.

### Objetivo 1

* Implementar la tool que:

  - obtiene IDs de topstories.json,

  - consulta item/{id}.json por cada historia,

  - elimina text si existe,

- devuelve title, by, time, url (con fallback a news.ycombinator.com/item?id=... si no hay url).

* Maneja timeouts y errores.

### Objetivo 2

* Implementa un caché en cache/ que:

  - guarde el último lote de historias “buenas”,

  - permita “modo offline” si la red falla o si se pasa --offline.

* Documenta la política de expiración (p. ej., 15 min).

### Objetivo 3

* Crea agente LangChain con:

  - ChatOpenAI(gpt-4, streaming=True, callbacks=[StreamingStdOutCallbackHandler()]),

  - ConversationBufferMemory.

* Prompt del editor: “Dame 10 titulares y sintetiza 3 disrupciones clave”.

* El agente debe usar la tool para extraer datos, no “inventarlos”.

### Objetivo 4

* Añade parámetros opcionales:

--min_score, --since_days, --only_domains=….

* Filtra o reordena historias antes de pasar al LLM.

* Expón el set final al LLM con metadatos (autor, score, dominio) para mejorar la síntesis.

### Objetivo 5

* Añade spans OTLP (descargables o a LangSmith).

* Crea tests mínimos:

- get_top_hn(5) retorna lista de dicts con claves requeridas,

  --offline usa caché,

- filtros respetados.
