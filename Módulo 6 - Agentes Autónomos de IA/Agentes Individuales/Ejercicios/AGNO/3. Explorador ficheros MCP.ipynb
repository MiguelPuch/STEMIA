# Explorador de ficheros vía MCP

## Objetivo

Crear un agente asíncrono que navegue el sistema de ficheros (sólo rutas permitidas) usando MCPTools y responda preguntas sobre el repo. Sin resolver.

## Requisitos

- Python 3.10+

- Paquetes: agno, python-dotenv

- Node (para npx @modelcontextprotocol/server-filesystem)

- Variables en .env: OPENAI_API_KEY

## Objetivos

### Objetivo 1

* Preparar MCPTools con comando:
npx -y @modelcontextprotocol/server-filesystem <ruta_base_permitida>

* Conectar (await mcp_tools.connect()), y cerrar al final.


### Objetivo 2

* Crear Agent con OpenAIChat(id="gpt-4o-mini").

* Instrucciones del sistema:

     - “Navega directorios permitidos”

     - “Usa list_allowed_directories para descubrir rutas”

     - “Sé conciso; usa markdown y títulos”

* Integrar mcp_tools como tool.

### Objetivo 3

* Implementar prompts de ejemplo (por consola):

        - “¿Qué licencia tiene este proyecto?”

        - “Enséñame los README que encuentres”

        - “Lista los 5 archivos más grandes”

* Para cada prompt, el agente debe planear y usar la(s) tool(s) MCP adecuadas (listar, leer fragmentos, etc.), y componer respuesta.

### Objetivo 4

* Manejo de errores (timeout del servidor MCP, permisos).

* Añadir bandera --base <ruta> para elegir raíz en ejecución.

* Añadir tests mínimos (p. ej., que list_allowed_directories devuelve algo y que la respuesta del agente incluye un encabezado markdown).
