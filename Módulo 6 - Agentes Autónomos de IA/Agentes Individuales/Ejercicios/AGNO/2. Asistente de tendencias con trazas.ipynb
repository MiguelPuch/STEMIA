# Analista de tendencias con trazas (OTLP) + Hacker News

### Objetivo

Crear un analista que agregue noticias (vía tool) y emita trazas con OpenTelemetry a un endpoint OTLP (p. ej. LangSmith o un colector local). Sin resolver.

### Requisitos

Paquetes: httpx, opentelemetry-sdk, opentelemetry-exporter-otlp-proto-http, python-dotenv, agno

Variables en .env: OPENAI_API_KEY, OTLP_ENDPOINT, LANGSMITH_API_KEY (si procede), LANGSMITH_PROJECT

### Objetivos

#### Objetivo 1

* Implementar una tool get_top_hn(num_stories: int) -> list[dict] que consulte:

https://hacker-news.firebaseio.com/v0/topstories.json

y cada item/{id}.json (sin el campo text si viene).

* Devolver una lista con title, by, time, url.

#### Objetivo 2

* Crear Agent con instrucciones tipo: “Eres periodista tecnológico; sintetiza disrupciones a partir de titulares.”

* Añadir la tool get_top_hn.

* Prompt de usuario: “¿Qué disrupciones prevés para el próximo trimestre?”

#### Objetivo 3

* Instrumentar agno (usa AgnoInstrumentor().instrument()).

* Añadir spans manuales alrededor de:

- llamada a HN,

- síntesis del modelo.

#### Objetivo 4

* Añadir un hook de confirmación que pida permiso si num_stories > 20.

* Si el usuario dice “n”, abortar educadamente.

#### Objetivo 5

* Documentar cómo arrancar con/ sin OTLP (p. ej. variable de entorno opcional).

* Añadir .env.example.

* Añadir requirements.txt.
