# Práctica 2 - Vigilancia de marca y reputación con sistemas multiagente

Monitorización automática de reputación online mediante agentes inteligentes

### Contexto:

Tu cliente (una escuela tecnológica) necesita un sistema que vigile menciones y noticias sobre sus cursos en IA y Data Science en España y Latinoamérica.
El sistema debe identificar oportunidades, riesgos reputacionales y temas emergentes.

### Objetivo:

Construir un pipeline multiagente que recopile información, la clasifique y genere un informe ejecutivo.

### Roles mínimos:

* Crawler: recopila resultados web relevantes.

* Classifier: determina si los resultados son positivos, negativos o neutros.

* Summarizer: agrupa los hallazgos en temas.

* Reporter: genera un informe final en Markdown con conclusiones y recomendaciones.

### Requisitos técnicos:

1. Definir los agentes con agno y un modelo como gpt-4o-mini.

2. Utilizar TavilyTools para hacer búsquedas web.

3. Incluir consultas en un archivo queries.txt.

4. El resultado final debe generarse en output/informe.md con tablas, citas y puntuación de sentimiento.
