# Práctica 1 – Equipo SEO multiagente

Optimización de contenidos web con un equipo multiagente de marketing digital

### Contexto:

Una escuela de formación profesional en Data Science e Inteligencia Artificial (como The Bridge) quiere mejorar el posicionamiento SEO de su web.
Para ello, implementará un sistema multiagente donde distintos “perfiles” colaboran para analizar el mercado, crear contenido y auditar la calidad final.

### Objetivo:

Diseñar y ejecutar en Python un equipo multiagente que simule un equipo de marketing digital formado por varios agentes especializados que trabajen juntos:

* Agente SEO Analyst: investiga palabras clave y competidores.

* Agente Content Strategist: redacta un texto atractivo y optimizado.

* Agente QA Evaluator: revisa la calidad, tono y estructura del contenido.

* Supervisor o Coordinador: coordina la interacción entre los agentes.

Cada agente debe tener su propio rol, instrucciones y herramientas (por ejemplo, una herramienta de búsqueda o razonamiento).

### Requisitos mínimos:

* Usar la librería agno para definir y ejecutar los agentes.

* Incorporar variables de entorno mediante dotenv.

* Mostrar el resultado final en un archivo output/draft.md con el texto consolidado y referencias.

* Documentar cómo se comunica cada agente con los demás.
